# Entrenamiento del alfabeto estático (Google Colab)

Equivale a `ml/train_static.py` (§10.5) para cuando TensorFlow no instala en el computador (por ejemplo en Windows, D-017).
Usa **uv** con el mismo `ml/pyproject.toml` del kit, así las versiones son idénticas (TensorFlow 2.15.1 y tensorflowjs 4.22.0).

1. Sube `ml/pyproject.toml`, `ml/train_static.py`, `content/signs.csv` y los `dataset.json` del colector (para los números de LSC-54: `dataset-lsc54-numeros.json`, D-019).
2. Ejecuta las celdas en orden.
3. Descarga `lsc-alfabeto.zip` y copia la carpeta **tal cual** en `apps/web/public/models/lsc-alfabeto/` (no renombres `group1-shard*.bin`).

```python
# uv suele venir en Colab; si no, se instala con su instalador oficial.
!command -v uv || curl -LsSf https://astral.sh/uv/install.sh | sh
import os
os.environ['PATH'] = os.path.expanduser('~/.local/bin') + ':' + os.environ['PATH']
!uv --version
```

```python
import os, shutil
from google.colab import files
print('Sube pyproject.toml, train_static.py, signs.csv y tus dataset*.json')
uploaded = files.upload()
for folder in ('ml/data', 'content'):
    os.makedirs(folder, exist_ok=True)
for name in uploaded:
    if name == 'pyproject.toml':
        shutil.move(name, 'ml/pyproject.toml')
    elif name == 'train_static.py':
        shutil.move(name, 'ml/train_static.py')
    elif name == 'signs.csv':
        shutil.move(name, 'content/signs.csv')
    elif name.endswith('.json'):
        shutil.move(name, f'ml/data/{name}')
print(sorted(os.listdir('ml/data')))
```

```python
# uv crea el entorno con Python 3.11 y las versiones fijas de ml/pyproject.toml, y entrena.
!MPLBACKEND=Agg uv run --project ml python ml/train_static.py --data 'ml/data/*.json' --out lsc-alfabeto --signs content/signs.csv --reports ml/reports
```
